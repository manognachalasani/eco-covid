```python
from pathlib import Path

# Load the final dataset
data_path = Path("c:/Users/roxan/OneDrive/Documents/DigitalLab/final_dataset.csv")

df = pd.read_csv(data_path)

# Display the first 5 rows
display(df.head())
```

```python
# Count missing values
missing_values = df.isnull().sum()

# Calculate missing percentage
missing_percentage = df.isnull().mean() * 100

# Create summary table
missing_summary = pd.DataFrame({
    "Missing values": missing_values,
    "Missing (%)": missing_percentage
})

# Sort by missing percentage
missing_summary = missing_summary.sort_values(
    by="Missing (%)",
    ascending=False
)

display(missing_summary)
```

```python
# Descriptive statistics
display(df.describe())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove rows without country or date
df_clean = df_clean.dropna(subset=["country", "date"])

# Remove completely empty columns
df_clean = df_clean.dropna(axis=1, how="all")

# Convert year to integer
df_clean["year"] = df_clean["year"].astype(int)

# Check the result
print("Dataset shape:", df_clean.shape)
```

```python
# Key indicators by analytical chapter
chapter_indicators = {
    "Pandemic & Policy": [
        "new_cases",
        "deaths_per_million",
        "stringency_index",
        "vaccinated_per_hundred"
    ],
    "Economic Impact": [
        "wb_gdp_growth",
        "wb_inflation",
        "gdp_shock_2020"
    ],
    "Employment": [
        "wb_unemployment_rate",
        "unemployment_change_2020"
    ],
    "Country Explorer": [
        "wb_gdp_per_capita",
        "human_development_index",
        "cum_deaths_per_million"
    ],
    "Recovery": [
        "wb_gdp_usd",
        "wb_gdp_growth"
    ]
}

# Create a data availability table
rows = []

for chapter, indicators in chapter_indicators.items():
    for indicator in indicators:
        for country, group in df_clean.groupby("country"):
            rows.append({
                "Chapter": chapter,
                "Indicator": indicator,
                "Country": country,
                "Available observations": group[indicator].notna().sum(),
                "Total observations": len(group)
            })

availability = pd.DataFrame(rows)

display(availability)
```

```python
availability_table = availability.pivot_table(
    index=["Chapter", "Indicator"],
    columns="Country",
    values="Available observations"
)

display(availability_table)
```

```python
# Chapter 1 — Pandemic & Policy Response

pandemic_columns = [
    "country",
    "date",
    "new_cases",
    "new_deaths",
    "cases_per_million",
    "deaths_per_million",
    "stringency_index",
    "vaccinated_per_hundred"
]

df_pandemic = df_clean[pandemic_columns].copy()

# Convert dates to datetime
df_pandemic["date"] = pd.to_datetime(
    df_pandemic["date"],
    dayfirst=True,
    errors="coerce"
)

# Keep observations from 2020 onwards
df_pandemic = df_pandemic[
    df_pandemic["date"].dt.year >= 2020
]

print("Pandemic dataset shape:", df_pandemic.shape)

display(df_pandemic.head())
```

```python
# Check actual date coverage for each indicator
for indicator in [
    "deaths_per_million",
    "stringency_index",
    "vaccinated_per_hundred"
]:
    print(f"\n{indicator}")

    coverage = (
        df_pandemic
        .dropna(subset=[indicator])
        .groupby("country")
        .agg(
            first_date=("date", "min"),
            last_date=("date", "max"),
            observations=(indicator, "count")
        )
    )

    display(coverage)
```

```python
import plotly.express as px

# Select available data
mortality_data = df_pandemic.dropna(
    subset=["date", "deaths_per_million"]
).copy()

# Create the line chart
fig = px.line(
    mortality_data,
    x="date",
    y="deaths_per_million",
    color="country",
    title="Reported COVID-19 Deaths per Million",
    labels={
        "date": "Date",
        "deaths_per_million": "Monthly reported deaths per million",
        "country": "Country"
    }
)

fig.update_layout(
    template="plotly_white",
    hovermode="x unified",
    legend_title_text="Country"
)

fig.show()
```

```python
# Choose three countries for a clearer comparison
selected_countries = [
    "United States",
    "United Kingdom",
    "China"
]

# Filter data
mortality_filtered = mortality_data[
    mortality_data["country"].isin(selected_countries)
]

# Create figure
fig = px.line(
    mortality_filtered,
    x="date",
    y="deaths_per_million",
    color="country",
    title="COVID-19 Mortality: Contrasting Pandemic Waves",
    labels={
        "date": "Date",
        "deaths_per_million": "Reported monthly deaths per million",
        "country": "Country"
    }
)

# Improve visual appearance
fig.update_layout(
    template="plotly_white",
    hovermode="x unified",
    height=500,
    legend=dict(
        orientation="h",
        y=-0.25,
        x=0.5,
        xanchor="center"
    )
)

fig.show()
```
